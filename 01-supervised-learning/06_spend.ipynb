## Add the boosting model

Build `hgb_pipeline` with `make_candidate_pipeline` and `HistGradientBoostingRegressor(random_state=42)`. Use `task="spend"`.

`hgb_grid` searches `model__learning_rate` at `0.05` and `0.1`, and `model__max_depth` at `3` and `6`.

The next cell fits that grid on the train rows and prints the comparison. A mean `DummyRegressor` is printed as a reference, not as a candidate. Its validation MAE is near 1229 THB. Expect ridge near 1045 THB. The forest and the boosting model are a little higher. Each candidate should beat the mean. Pick the lowest validation MAE.


```python
from sklearn.ensemble import HistGradientBoostingRegressor

from pipelines import make_candidate_pipeline

hgb_pipeline = None
hgb_grid = None
```

```python
import time

from sklearn.model_selection import GridSearchCV

from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
from sklearn.model_selection import cross_val_score

from pipelines import format_comparison, score_row, spend_folds
from spend import clip_spend

if hgb_pipeline is None or hgb_grid is None:
    raise NotImplementedError("Build the boosting pipeline and its grid.")
started = time.perf_counter()
hgb_search = GridSearchCV(
    hgb_pipeline,
    hgb_grid,
    scoring="neg_mean_absolute_error",
    cv=spend_folds(),
    refit=True,
    n_jobs=1,
)
hgb_search.fit(X_train, y_train_spend)
hgb_seconds = time.perf_counter() - started
rows = [
    score_row("ridge", ridge_search, ridge_seconds, X_val, y_val_spend, "spend"),
    score_row("forest", forest_search, forest_seconds, X_val, y_val_spend, "spend"),
    score_row("hist_gradient", hgb_search, hgb_seconds, X_val, y_val_spend, "spend"),
]
print(format_comparison(rows, "spend"))
mean_model = DummyRegressor(strategy="mean")
mean_cv = -cross_val_score(
    mean_model, X_train, y_train_spend, cv=spend_folds(), scoring="neg_mean_absolute_error", n_jobs=1
)
mean_model.fit(X_train, y_train_spend)
mean_pred = clip_spend(mean_model.predict(X_val))
mean_mae = float(mean_absolute_error(y_val_spend, mean_pred))
mean_rmse = float(root_mean_squared_error(y_val_spend, mean_pred))
print(
    f"{'mean_baseline':<16}{mean_cv.mean():10.1f}{mean_cv.std():10.1f}{mean_mae:10.1f}{mean_rmse:10.1f}"
)
print("mean_baseline is a reference. It is not a candidate.")
```
